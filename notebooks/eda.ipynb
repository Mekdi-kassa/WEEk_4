## 1. Setup and Data Loading

- Load required libraries.
- Read raw transaction data from data/raw/.
- Display first rows and schema.

```python
import warnings

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

DATA_PATH = "../data/raw/data.csv"

df = pd.read_csv(DATA_PATH)
df.head()
```

## 2. Data Overview

Check shape, data types, uniqueness, and a quick profile of columns.

```python
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

print("\nUnique values per column:")
print(df.nunique().sort_values(ascending=False))
```

## 3. Summary Statistics

```python
df.describe(include="all").T
```

## 4. Numerical Feature Distributions

```python
num_cols = df.select_dtypes(include=["number"]).columns.tolist()

if num_cols:
    df[num_cols].hist(figsize=(14, 10), bins=30)
    plt.suptitle("Numerical Feature Distributions")
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found.")
```

## 5. Categorical Feature Distributions

```python
cat_cols = df.select_dtypes(exclude=["number"]).columns.tolist()

for col in cat_cols:
    plt.figure(figsize=(12, 4))
    top_values = df[col].astype(str).value_counts().head(15)
    sns.barplot(x=top_values.index, y=top_values.values)
    plt.title(f"Top categories in {col}")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

## 6. Correlation Analysis

```python
if len(num_cols) > 1:
    corr = df[num_cols].corr(numeric_only=True)
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, cmap="coolwarm", center=0)
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()
else:
    print("Not enough numerical columns for correlation analysis.")
```

## 7. Missing Values

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
missing_report = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
missing_report[missing_report["missing_count"] > 0]
```

## 8. Outlier Detection

```python
for col in num_cols:
    plt.figure(figsize=(10, 2.8))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot: {col}")
    plt.tight_layout()
    plt.show()
```
